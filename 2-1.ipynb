## 훈련 세트와 테스트 세트

```python
beak_length = [8.2, 8.5, 8.7, 9.0, 9.1, 9.3, 9.5, 9.7, 9.8, 10.0, 8.8, 9.2, 9.6, 10.2, 8.6, 9.4,
    10.1, 8.9, 9.9, 10.3, 8.4, 9.0, 9.7, 10.0, 8.7, 9.5, 10.4, 9.3, 8.9, 9.8, 10.1, 9.2,
    8.5, 9.6, 10.2, 12.5, 13.0, 13.4, 13.8, 14.0, 14.3, 14.5, 14.8,
    15.0, 15.3, 12.8, 13.6, 14.2, 15.5]

beak_height = [
    5.8, 6.1, 5.9, 6.3, 6.0, 6.5, 6.2, 6.7,
    6.4, 6.8, 6.1, 6.4, 6.0, 6.6, 5.7, 6.3,
    6.9, 6.2, 6.5, 6.7, 5.9, 6.1, 6.3, 6.4,
    6.0, 6.6, 6.8, 6.2, 5.8, 6.7, 6.5, 6.0,
    5.6, 6.1, 6.9, 3.2, 3.0, 3.3, 2.9, 3.1, 2.8, 3.2, 2.7,
    3.0, 2.9, 3.1, 2.8, 3.0, 2.7
]
```

```python
beak_data = [[l, h] for l, h in zip(beak_length, beak_height)]
beak_target = [1]*35 + [0]*14
```

```python
from sklearn.neighbors import KNeighborsClassifier

kn = KNeighborsClassifier()
```

```python
print(beak_data[4])
```

```python
print(beak_data[0:5])
```

```python
print(beak_data[:5])
```

```python
print(beak_data[44:])
```

```python
train_input = beak_data[:35]
train_target = beak_target[:35]

test_input = beak_data[35:]
test_target = beak_target[35:]
```

```python
kn.fit(train_input, train_target)
kn.score(test_input, test_target)
```
