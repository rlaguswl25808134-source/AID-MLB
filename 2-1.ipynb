## 훈련 세트와 테스트 세트

```python
beak_length = [8.2, 8.5, 8.7, 9.0, 9.1, 9.3, 9.5, 9.7, 9.8, 10.0, 8.8, 9.2, 9.6, 10.2, 8.6, 9.4,
    10.1, 8.9, 9.9, 10.3, 8.4, 9.0, 9.7, 10.0, 8.7, 9.5, 10.4, 9.3, 8.9, 9.8, 10.1, 9.2,
    8.5, 9.6, 10.2, 12.5, 13.0, 13.4, 13.8, 14.0, 14.3, 14.5, 14.8,
    15.0, 15.3, 12.8, 13.6, 14.2, 15.5]

beak_height = [
    5800, 6100, 5900, 6300, 6000, 6500, 6200, 6700,
    6400, 6800, 6100, 6400, 6000, 6600, 5700, 6300,
    6900, 6200, 6500, 6700, 5900, 6100, 6300, 6400,
    6000, 6600, 6800, 6200, 5800, 6700, 6500, 6000,
    5600, 6100, 6900,

    3200, 3000, 3300, 2900, 3100, 2800, 3200, 2700,
    3000, 2900, 3100, 2800, 3000, 2700
]
```

```python
beak_data = [[l, h] for l, h in zip(beak_length, beak_height)]
beak_target = [1]*35 + [0]*14
```

```python
from sklearn.neighbors import KNeighborsClassifier

kn = KNeighborsClassifier()
```

```python
kn.fit(train_input, train_target)
kn.score(test_input, test_target)
```
